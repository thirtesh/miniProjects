```python
import os
import joblib as jb
import warnings
warnings.filterwarnings('ignore')
```

```python
file=os.path.expanduser('~/Desktop/Project/heart_disease/Model/model.joblib')
model=jb.load(file)
```

```python
print("Enter the following details:-")
age=int(input("Age:"))
sex=int(input("Gender(1-male;0-female):"))	
cp=int(input("Chest pain type(0-3):"))
trestbps=int(input("Is Resting blood pressure:"))
chol=int(input("Cholestrol level:"))
fbs=int(input("Fasting blood sugar level >120mg/dl:"))
restecg=int(input("Resting ECG result:"))
thalach=int(input("Maximum Heart Rate:"))
exang=int(input("Exercise induced Angina(1-yes;0-no):"))
oldpeak=float(input("ST depression:"))
slope=int(input("Slope of Peak Excercise:"))
ca=int(input("Number of Major Vessels:"))
thal=int(input("Thallium Stress Result:"))
```

```python
pred=model.predict([[age, sex,cp,trestbps,chol,fbs,restecg,thalach,exang,oldpeak,slope,ca,thal]])[0]
if pred==0:
    print("You DO NOT have any Heart Disease")
else:
    print("You DO have some Heart Disease")
```
