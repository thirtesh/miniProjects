```python
import joblib as jb
import os
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
file= os.path.expanduser('~/Desktop/Project/wine_quality/Model/model.joblib')
model = jb.load(file)
```

```python
print("Enter the specifications of the wine:-\n")
f_acid=float(input("\n\tFixed acidity:"))
v_acid=float(input("\n\tVolatile acidity:"))
res_sug=float(input("\n\tResidual sugar:"))
cl=float(input("\n\tChlorides:"))
tsd=float(input("\n\tTotal Sulphur Dioxide:"))
dens=float(input("\n\tDensity:"))
ph=float(input('\n\tpH:'))
alc=float(input('\n\tAcohol:'))
```

```python
pred=model.predict(np.array([f_acid, v_acid, res_sug, cl, tsd, dens, ph, alc]).reshape(1,-1))[0]
if pred==1:
    print("The wine is GOOD")
else:
    print("The wine is NOT good")
```
