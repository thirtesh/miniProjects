```python
import pandas as pd
import sklearn as sk
```

```python
#importing and accessing dataset
data=pd.DataFrame(pd.read_csv("~/Downloads/iris.csv"))
data.head()
```

```python
#splitting data into X and y
X=data.drop(['species'], axis=1)
y=data['species']
display(X.head())
display(y.head())
```

```python
#splitting data for training and testing
X_train, X_test, y_train, y_test = sk.model_selection.train_test_split(X, y, test_size=0.2, random_state=77)
```

```python
#pipelining for preprocessing and model
y_train= sk.preprocessing.LabelEncoder().fit_transform(y_train)
y_test= sk.preprocessing.LabelEncoder().fit_transform(y_test)
estimators=[('stdScaler',sk.preprocessing.StandardScaler()), ('KNN',sk.neighbors.KNeighborsClassifier())]
pipe = sk.pipeline.Pipeline(estimators)
pipe.get_params()
```

```python
#parameter gird and GridSearchCV
param_grid={'stdScaler__with_mean':[True,False], 'KNN__n_neighbors':[x for x in range(1,10)], 'KNN__weights':['uniform','distance']}
grid= sk.model_selection.GridSearchCV(pipe, param_grid, cv=5, scoring='accuracy')
```

```python
#fitting
grid.fit(X_train, y_train)
```

```python
#scores(separated from fitting so that its easy to re-run)
display(grid.best_params_)
display(grid.score(X_test,y_test))
```
