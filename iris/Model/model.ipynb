```python
import joblib as jb
import os
import sklearn as sk
import pandas as pd
```

```python
#import 
iris_names=['Iris Setosa', 'Iris Vesicolor', 'Iris Virginica']
file=os.path.expanduser("~/Desktop/Project/iris/Model/model.joblib")
model=jb.load(file)
```

```python
#get user input(kindly enter the values if you're testing the model)
sl=float(input('Enter the Sepal Length:'))
sw=float(input('Enter the Sepal Width:'))
pl=float(input('Enter the Petal Length:'))
pw=float(input('Enter the Petal Width:'))
```

```python
#predict using the model and give tyhe output
data=pd.DataFrame([[sl,sw,pl,pw]],columns=['sepal_length','sepal_width','petal_length','petal_width'])
res=model.predict(data)
print(f'It\'s likely an {iris_names[res[0]]}')
```
